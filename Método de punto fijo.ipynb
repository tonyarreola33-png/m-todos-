MÉTODO DE PUNTO FIJO

Este codigo tiene el **Método de Iteración de Punto Fijo**  
El método de punto fijo busca encontrar una solución a la ecuación:
$$p = g(p)$$

Dada una aproximación inicial $p_0$, generamos una sucesión $\{p_n\}_{n=0}^{\infty}$ mediante la relación de recurrencia:
$$p_i = g(p_{i-1}), \quad \text{para } i \geq 1$$

Si la función $g$ es continua y la sucesión converge, entonces el límite es un punto fijo de $g$.

El problema consiste en encontrar la raíz de la ecuación:
$$f(x) = x^3 + 4x^2 - 10 = 0$$

```python

import numpy as np
import matplotlib.pyplot as plt
```

```python

def f(x):
    return x**3+4*x**2-10
```

```python

xx=np.linspace(1,2)
plt.plot(xx,f(xx))
plt.axhline(y=0, color='green')
plt.axvline(x=0, color='red')
```

```python

def g(x):
    return .05*np.sqrt(10-x**3)
```

```python
g(1.5)
```

```python

def pfijo(p0,i):

   p0=1.5
   lista=[]
   error=[]

   for i in range(10):
      p1=g(p0)
      lista.append(p1)
      e=np.abs(p1-p0)
      error.append(e)
      p0=p1
   return lista, error
```

```python
lista, error =pfijo(1.5,15)
```
